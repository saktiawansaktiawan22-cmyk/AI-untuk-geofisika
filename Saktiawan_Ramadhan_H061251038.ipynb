# Prompt 1 — Interpretasi Data Seismik Refraksi

```markdown
# R — Role
Anda berperan sebagai ahli seismik refraksi yang membimbing mahasiswa S1 Geofisika dalam menginterpretasi data waktu tempuh gelombang seismik.

# T — Task
Interpretasikan data waktu tiba pertama (first arrival) untuk menentukan kecepatan gelombang pada tiap lapisan dan perkiraan ketebalan lapisan di bawah permukaan.

# C — Context
Survei seismik refraksi dangkal digunakan untuk memperkirakan struktur lapisan bawah permukaan, misalnya batas antara lapisan lapuk dan batuan yang lebih kompak. Analisis dilakukan dalam mata kuliah Metode Seismik Refraksi sebagai latihan interpretasi kurva waktu–jarak (time–distance).

# I — Input
Gunakan data yang diberikan berupa jarak geophone dari titik tembak (offset) dan waktu tiba pertama gelombang. Perhatikan satuan jarak dan waktu yang tertera pada data.

# K — Constraints
- Jangan mengubah nilai asli data.
- Jelaskan bagaimana titik-titik data dipisahkan menjadi segmen gelombang langsung dan gelombang bias.
- Jelaskan asumsi model yang digunakan, misalnya lapisan horizontal dan kecepatan meningkat terhadap kedalaman.
- Jika ada titik data yang tampak menyimpang, jelaskan dan jangan dibuang tanpa alasan.
- Jangan menyimpulkan jenis batuan hanya dari nilai kecepatan.

# O — Output
Hasil harus mencakup:
1. Deskripsi singkat data dan satuannya.
2. Cara pembacaan kurva waktu–jarak dan pembagian segmennya.
3. Kecepatan gelombang tiap lapisan beserta cara memperolehnya.
4. Perkiraan ketebalan lapisan pertama beserta langkah perhitungannya.
5. Interpretasi geologi secara hati-hati dalam bentuk paragraf.
6. Keterbatasan hasil dan kesimpulan singkat.

# Q — Quality Criteria
Perhitungan harus benar dan satuannya konsisten, langkah interpretasi dapat diikuti mahasiswa S1, interpretasi tidak melebihi apa yang didukung data, dan hasil dapat digunakan langsung dalam laporan praktikum.

# V — Verification
Periksa kembali apakah kecepatan lapisan bawah lebih besar dari lapisan atas, apakah titik potong (crossover) kedua segmen sesuai dengan kurva, dan apakah hasil perhitungan masuk akal dibandingkan nilai kecepatan tipikal material dangkal.
```

# Prompt 2 — Analisis Katalog Gempa

```markdown
# R — Role
Anda berperan sebagai seismolog yang membimbing mahasiswa S1 Geofisika dalam menganalisis data kejadian gempa bumi.

# T — Task
Analisis katalog gempa untuk mengetahui sebaran spasial, sebaran kedalaman, distribusi magnitudo, dan pola kejadian gempa pada wilayah studi.

# C — Context
Katalog gempa memuat waktu kejadian, lokasi, kedalaman, dan magnitudo. Analisis dilakukan dalam mata kuliah Seismologi untuk memahami karakteristik seismisitas suatu wilayah dan kaitannya dengan kondisi tektonik, misalnya zona subduksi atau sesar aktif.

# I — Input
Gunakan katalog gempa yang diberikan, termasuk waktu kejadian, koordinat episenter, kedalaman, magnitudo, dan jenis magnitudo. Gunakan hanya data yang tersedia pada katalog.

# K — Constraints
- Jangan mengubah nilai asli data.
- Jangan menyamakan jenis magnitudo yang berbeda tanpa penjelasan.
- Jelaskan bagaimana data kosong atau data di luar wilayah studi ditangani.
- Jangan membuat prediksi gempa atau penilaian bahaya gempa.
- Kaitan dengan tektonik hanya disampaikan sebagai dugaan yang dijelaskan dasarnya.

# O — Output
Hasil harus mencakup:
1. Deskripsi katalog: periode, wilayah, dan jumlah kejadian.
2. Statistik dasar magnitudo dan kedalaman.
3. Pola sebaran episenter dan kedalaman.
4. Pola kejadian terhadap waktu.
5. Interpretasi hasil dalam bentuk paragraf.
6. Kesimpulan singkat.

# Q — Quality Criteria
Angka harus sesuai katalog, istilah seismologi dipakai dengan tepat, interpretasi tektonik tidak berlebihan, dan hasil dapat digunakan langsung sebagai bagian dari laporan Seismologi.

# V — Verification
Periksa kembali jumlah kejadian yang dianalisis, rentang magnitudo dan kedalaman yang dilaporkan, serta pastikan setiap kesimpulan dapat ditelusuri ke data pada katalog.
```

# Prompt 3 — Analisis Geostatistika: Variogram dan Interpolasi

```markdown
# R — Role
Anda berperan sebagai ahli geostatistika yang membimbing mahasiswa S1 Geofisika dalam memahami ketergantungan spasial dan melakukan interpolasi data.

# T — Task
Analisis keterkaitan spasial nilai data pada titik-titik pengukuran menggunakan variogram, lalu jelaskan bagaimana hasilnya dipakai untuk interpolasi kriging dan seberapa andal hasil interpolasinya.

# C — Context
Data geofisika hanya diukur pada titik tertentu, sedangkan peta sebaran membutuhkan nilai di seluruh area. Analisis dilakukan dalam mata kuliah Geostatistika dan Analisis Spasial untuk memahami sejauh mana titik yang berdekatan saling mirip, sebelum melakukan interpolasi.

# I — Input
Gunakan dataset yang diberikan berupa koordinat titik pengukuran dan satu variabel geofisika, misalnya anomali gravitasi, ketebalan lapisan, atau kedalaman muka air tanah. Perhatikan sistem koordinat dan satuan jarak pada data.

# K — Constraints
- Jangan mengubah nilai asli data tanpa alasan yang jelas.
- Pastikan koordinat berada pada sistem yang satuannya meter sebelum menghitung jarak.
- Jangan membuang data pencilan (outlier) secara diam-diam; jelaskan pencilan dan pengaruhnya.
- Jelaskan alasan pemilihan jarak antar-pasangan titik (lag) dan model variogram.
- Jangan menafsirkan nilai di luar sebaran titik data seolah-olah sama andalnya.

# O — Output
Hasil harus mencakup:
1. Deskripsi dataset dan statistik dasar.
2. Penjelasan variogram eksperimental dan cara membacanya.
3. Arti nugget, sill, dan range pada data ini.
4. Model variogram yang dipilih beserta alasannya.
5. Penjelasan hasil kriging dan tingkat ketidakpastiannya.
6. Perbandingan singkat dengan metode interpolasi yang lebih sederhana.
7. Kesimpulan singkat.

# Q — Quality Criteria
Konsep geostatistika dijelaskan benar dan mudah dipahami, nilai parameter disebutkan dengan satuannya, keterbatasan data (jumlah dan sebaran titik) dibahas, dan hasil dapat dipakai langsung dalam laporan praktikum.

# V — Verification
Periksa apakah nilai range masuk akal terhadap ukuran area, apakah interpolasi diuji dengan cara meninggalkan satu titik lalu memprediksinya kembali (leave-one-out), dan apakah kesimpulan sesuai dengan hasil pengujian tersebut.
```

# Prompt 4 — Model Pendinginan Litosfer

```markdown
# R — Role
Anda berperan sebagai ahli geodinamika dan geofisika komputasi yang membimbing mahasiswa S1 Geofisika dalam memahami pendinginan litosfer samudra.

# T — Task
Jelaskan dan analisis model sederhana pendinginan litosfer satu dimensi, lalu bandingkan hasil hitungan numerik dengan solusi analitiknya untuk beberapa umur lempeng.

# C — Context
Litosfer samudra menebal seiring bertambahnya umur karena kehilangan panas ke permukaan. Analisis dilakukan dalam mata kuliah Geodinamika dan Tektonofisika serta Geofisika Komputasi untuk memahami hubungan antara umur lempeng, profil suhu, dan ketebalan litosfer.

# I — Input
Gunakan parameter yang diberikan, yaitu difusivitas termal, suhu permukaan, suhu mantel, kedalaman maksimum, dan beberapa nilai umur lempeng. Tidak ada data lain yang boleh dipakai.

# K — Constraints
- Jangan mengubah parameter yang diberikan tanpa menyebutkannya.
- Jelaskan asumsi model, termasuk hal yang tidak dimodelkan seperti pembangkitan panas dan konveksi.
- Pastikan semua satuan konsisten dan jelaskan konversinya, terutama untuk umur dalam juta tahun.
- Jelaskan syarat agar hitungan numerik stabil.
- Jangan menyimpulkan kondisi lempeng nyata dari model yang sangat sederhana.

# O — Output
Hasil harus mencakup:
1. Penjelasan persamaan dan asumsi model dalam bahasa yang mudah dipahami.
2. Kondisi awal dan kondisi batas yang dipakai.
3. Profil suhu terhadap kedalaman untuk tiap umur.
4. Perbandingan hasil numerik dan analitik beserta besar selisihnya.
5. Kedalaman isoterm acuan untuk tiap umur dan hubungannya dengan umur.
6. Interpretasi geodinamika dalam bentuk paragraf.
7. Keterbatasan model dan kesimpulan singkat.

# Q — Quality Criteria
Penjelasan fisis benar, selisih numerik dan analitik dinyatakan dengan angka (bukan hanya "mirip"), satuan konsisten, dan interpretasi tidak melampaui asumsi model.

# V — Verification
Periksa apakah selisih hasil numerik mengecil ketika resolusi kedalaman diperhalus, apakah suhu di permukaan dan di kedalaman maksimum sesuai kondisi batas, dan apakah isoterm bergeser lebih dalam untuk umur yang lebih tua.
```

# Prompt 5 — Klasifikasi Litologi dengan Machine Learning

```markdown
# R — Role
Anda berperan sebagai ahli AI untuk geofisika yang membimbing mahasiswa S1 Geofisika dalam menerapkan machine learning pada data log sumur.

# T — Task
Jelaskan dan evaluasi pendekatan machine learning untuk mengklasifikasikan jenis litologi dari data log sumur, termasuk perbandingan dengan model pembanding yang sederhana.

# C — Context
Data log sumur seperti gamma ray, densitas, porositas neutron, dan sonik dapat membantu membedakan litologi. Analisis dilakukan dalam mata kuliah AI untuk Geofisika untuk memahami kelebihan dan batas penerapan machine learning dalam interpretasi geologi.

# I — Input
Gunakan dataset yang diberikan berisi nama sumur, kedalaman, nilai log, dan label litologi. Perhatikan satuan tiap log dan asal label litologi.

# K — Constraints
- Jangan mengubah nilai asli data tanpa alasan yang jelas.
- Pemisahan data latih dan data uji harus dilakukan per sumur, bukan acak per baris, dan jelaskan alasannya.
- Jangan menggunakan data uji untuk menentukan pengaturan model.
- Jangan memakai kedalaman atau nama sumur sebagai fitur.
- Jelaskan penanganan data kosong dan nilai yang tidak masuk akal secara fisis.
- Jangan mengklaim model dapat digeneralisasi ke lapangan lain.

# O — Output
Hasil harus mencakup:
1. Deskripsi dataset dan sebaran kelas litologi.
2. Penjelasan model pembanding sederhana dan model machine learning yang dipilih.
3. Hasil evaluasi, termasuk performa tiap kelas litologi, bukan hanya akurasi total.
4. Analisis kesalahan: litologi mana yang sering tertukar dan mengapa secara geologi.
5. Bahasan pengaruh ketidakseimbangan kelas.
6. Interpretasi dalam bentuk paragraf dan kesimpulan singkat.

# Q — Quality Criteria
Evaluasi jujur dan tidak bocor (tanpa data leakage), performa dilaporkan per kelas, penjelasan mudah dipahami mahasiswa S1, dan hasil dapat dipakai langsung dalam laporan atau tugas.

# V — Verification
Periksa apakah tidak ada sumur yang muncul di data latih sekaligus data uji, apakah model machine learning benar-benar lebih baik daripada model pembanding, dan apakah performa turun ke tingkat kebetulan ketika label diacak.
```
