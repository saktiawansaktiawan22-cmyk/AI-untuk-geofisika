## Tujuan
Notebook ini digunakan untuk mengeksplorasi data gempa signifikan, melihat karakteristik magnitudo dan kedalaman, serta mengamati distribusi kejadian berdasarkan waktu dan lokasi.

Analisis yang dilakukan bersifat eksploratif (*Exploratory Data Analysis/EDA*). Dataset hanya berisi sejumlah kecil kejadian, sehingga hasilnya tidak digunakan untuk membangun model prediksi yang diklaim berlaku secara umum.


## 1. Import Library

Library yang digunakan adalah Pandas dan NumPy untuk pengolahan data, serta Matplotlib dan Seaborn untuk visualisasi.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)
```

## 2. Memuat Dataset

File dibaca dari folder `data`. Jika notebook dijalankan dari folder `notebooks`, path relatif berikut akan mengarah ke dataset di folder utama.

```python
df = pd.read_csv("../data/significant_month.csv")

print("Ukuran dataset:", df.shape)
display(df.head())
```

## 3. Memahami Struktur Data

Tahap ini menampilkan tipe data, informasi kolom, dan ringkasan statistik untuk mengetahui karakteristik awal dataset.

```python
df.info()
display(df.describe(include="all").T)
```

## 4. Pemeriksaan Data

Memeriksa nilai kosong dan data duplikat. Kolom waktu kemudian diubah ke format datetime agar dapat dianalisis.

```python
print("Jumlah nilai kosong per kolom:")
display(df.isnull().sum().sort_values(ascending=False))

print("Jumlah baris duplikat:", df.duplicated().sum())

df["time"] = pd.to_datetime(df["time"], errors="coerce")
df["mag"] = pd.to_numeric(df["mag"], errors="coerce")
df["depth"] = pd.to_numeric(df["depth"], errors="coerce")

data = df.dropna(subset=["time", "mag", "depth"]).copy()
print("Ukuran data setelah pembersihan:", data.shape)
```

## 5. Statistik Magnitudo dan Kedalaman

Magnitudo menggambarkan ukuran gempa, sedangkan kedalaman menunjukkan posisi sumber gempa di bawah permukaan bumi. Nilai statistik berikut memberikan gambaran umum dataset.

```python
print("Statistik magnitudo:")
display(data["mag"].describe())

print("\nStatistik kedalaman (km):")
display(data["depth"].describe())

print("\nRata-rata magnitudo:", round(data["mag"].mean(), 2))
print("Rata-rata kedalaman (km):", round(data["depth"].mean(), 2))
```

## 6. Distribusi Magnitudo

Histogram digunakan untuk melihat sebaran nilai magnitudo pada kejadian gempa yang tercatat.

```python
plt.figure()
sns.histplot(data=data, x="mag", bins=8, kde=True)
plt.title("Distribusi Magnitudo Gempa Signifikan")
plt.xlabel("Magnitudo")
plt.ylabel("Frekuensi")
plt.show()
```

## 7. Hubungan Magnitudo dan Kedalaman

Scatter plot memperlihatkan hubungan antara magnitudo dan kedalaman. Pola pada grafik merupakan gambaran dataset ini dan tidak langsung menunjukkan hubungan sebab-akibat.

```python
plt.figure()
sns.scatterplot(data=data, x="depth", y="mag", hue="magType", s=90)
plt.title("Hubungan Kedalaman dan Magnitudo Gempa")
plt.xlabel("Kedalaman (km)")
plt.ylabel("Magnitudo")
plt.legend(title="Tipe Magnitudo", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

## 8. Distribusi Tipe Magnitudo

Menghitung jumlah kejadian berdasarkan jenis pengukuran magnitudo (`magType`).

```python
plt.figure()
data["magType"].value_counts().plot(kind="bar")
plt.title("Jumlah Gempa Berdasarkan Tipe Magnitudo")
plt.xlabel("Tipe Magnitudo")
plt.ylabel("Jumlah Kejadian")
plt.xticks(rotation=0)
plt.show()
```

## 9. Distribusi Kejadian Berdasarkan Waktu

Data dikelompokkan berdasarkan tanggal untuk melihat jumlah kejadian yang tercatat pada setiap tanggal.

```python
data["tanggal"] = data["time"].dt.date
jumlah_harian = data.groupby("tanggal").size()

plt.figure(figsize=(12, 5))
jumlah_harian.plot(kind="bar")
plt.title("Jumlah Gempa Signifikan per Tanggal")
plt.xlabel("Tanggal")
plt.ylabel("Jumlah Kejadian")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## 10. Lokasi dengan Magnitudo Terbesar

Bagian ini menampilkan beberapa kejadian dengan magnitudo tertinggi berdasarkan kolom `place`.

```python
kolom_lokasi = ["place", "mag", "depth", "time"]
display(data.sort_values("mag", ascending=False)[kolom_lokasi].head(10))
```

## 11. Kesimpulan

Isi kesimpulan berdasarkan hasil yang muncul setelah seluruh sel dijalankan. Perhatikan nilai magnitudo tertinggi, rentang kedalaman, tipe magnitudo yang dominan, serta pola waktu dan lokasi.

**Catatan:** Dataset ini merupakan kumpulan kejadian gempa signifikan dalam periode tertentu dan hanya memiliki sedikit baris. Oleh karena itu, analisis ini lebih tepat digunakan untuk memahami karakteristik data daripada membuat generalisasi atau prediksi gempa di masa depan.
